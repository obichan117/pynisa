```python
import pynisa

print("Sources:", pynisa.sources())
print("Rakuten categories:", pynisa.categories("rakuten"))
print("SBI categories:", pynisa.categories("sbi"))
```

## JP Stock Rankings

```python
# Rakuten JP stock ranking (default)
result = pynisa.ranking("rakuten")
print(f"Period: {result.period}")
result.data
```

```python
# SBI JP stock ranking
result = pynisa.ranking("sbi")
print(f"Period: {result.period}")
result.data
```

## US Stock Rankings

```python
# Rakuten US stocks
result = pynisa.ranking("rakuten", "us_stock_buy")
result.data
```

```python
# SBI US stocks
result = pynisa.ranking("sbi", "us_stock_buy")
result.data
```

## Fund Rankings

```python
result = pynisa.ranking("sbi", "fund_buy")
result.data
```

## Compare Across Brokerages

```python
import pandas as pd

rakuten = pynisa.ranking("rakuten", "us_stock_buy", count=5)
sbi = pynisa.ranking("sbi", "us_stock_buy", count=5)

combined = pd.concat([rakuten.data, sbi.data], ignore_index=True)
combined[["rank", "ticker", "name", "source"]]
```
